```python
from sklearn.model_selection import train_test_split, GridSearchCV, KFold

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import StandardScaler, OneHotEncoder

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.feature_selection import SelectKBest, f_regression

from sklearn.tree import DecisionTreeRegressor

from sklearn.linear_model import LinearRegression

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import joblib
```

```python
df_clean = df.copy()

print("Original Dataset Shape:", df.shape)
print("Working Dataset Shape:", df_clean.shape)
```

```python
X = df_clean.drop('median_house_value', axis = 1)

y = df_clean['median_house_value']

print("Independent Variables Shape", X.shape)
print("Dependent Variable shape:", y.shape)
```

```python
X_train, X_test , y_train , y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42
)

print("Training Data:", X_train.shape)

print("Testing Data:", X_test.shape)

print("Training Target:", y_train.shape)

print("Testing Target:", y_test.shape)
```

```python
print("Missing Values Before Imputation:")

print(X_train.isnull().sum())
```

```python
numeric_columns = X_train.select_dtypes(
    include=np.number
).columns

outlier_counts = {}

for column in numeric_columns:

    Q1 = X_train[column].quantile(0.25)

    Q3 = X_train[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR

    upper_limit = Q3 + 1.5 * IQR

    outliers = (
        (X_train[column] < lower_limit) |
        (X_train[column] > upper_limit)
    )

    outlier_counts[column] = outliers.sum()

outlier_summary = pd.DataFrame(
    outlier_counts.items(),
    columns=['Feature', 'Outlier Count']
)

display(outlier_summary)
```

```python
plt.figure(figsize=(15, 10))

for i, column in enumerate(numeric_columns, 1):

    plt.subplot(3, 3, i)

    sns.boxplot(y=X_train[column])

    plt.title(column)

plt.tight_layout()

plt.show()
```

```python
def create_features(data):

    data = data.copy()

    data['rooms_per_household'] = (
        data['total_rooms'] / data['households'].replace(0, np.nan)
    )

    data['bedrooms_per_room'] = (
        data['total_bedrooms'] / data['total_rooms'].replace(0, np.nan)
    )

    data['population_per_household'] = (
        data['population'] / data['households'].replace(0, np.nan)
    )

    return data
```

```python
X_train = create_features(X_train)

X_test = create_features(X_test)

print("Training Shape:", X_train.shape)

print("Testing Shape:", X_test.shape)

display(X_train.head())
```

```python
X_train = create_features(X_train)

X_test = create_features(X_test)

print("Training Shape:", X_train.shape)

print("Testing Shape:", X_test.shape)

display(X_train.head())
```

```python
numeric_features = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include='object'
).columns.tolist()

print("Numerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
numeric_transformer_lr = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]
)
```

```python
categorical_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(
            handle_unknown='ignore',
            sparse_output=False
        ))
    ]
)
```

```python
preprocessor_lr = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_lr, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)
```

```python
numeric_transformer_dt = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='median'))
    ]
)
```

```python
preprocessor_dt = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_dt, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)
```

```python
# Feature Selection

feature_selector = SelectKBest(
    score_func=f_regression,
    k='all'
)
```

```python
X_train_processed = preprocessor_lr.fit_transform(X_train)

feature_names = preprocessor_lr.get_feature_names_out()

feature_selector.fit(
    X_train_processed,
    y_train
)

feature_scores = pd.DataFrame({
    'Feature': feature_names,
    'Score': feature_selector.scores_,
    'P_Value': feature_selector.pvalues_
})

feature_scores = feature_scores.sort_values(
    by='Score',
    ascending=False
)

display(feature_scores)
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=feature_scores.head(15),
    x='Score',
    y='Feature'
)

plt.title('Top 15 Features - Feature Selection')
plt.xlabel('F-Regression Score')
plt.ylabel('Features')

plt.show()
```

```python
selector = SelectKBest(
    score_func=f_regression,
    k=10
)

X_train_selected = selector.fit_transform(
    X_train_processed,
    y_train
)

selected_features = feature_names[selector.get_support()]

print("Selected Features:")

print(selected_features.tolist())

print("\nNumber of Selected Features:")

print(len(selected_features))
```

```python
# Building complete Machine earning Pipelines

decision_tree_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor_dt),

        ('feature_selection', SelectKBest(
            score_func=f_regression,
            k='all'
        )),

        ('model', DecisionTreeRegressor(
            random_state=42
        ))
    ]
)
```

```python
linear_regression_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor_lr),

        ('feature_selection', SelectKBest(
            score_func=f_regression,
            k='all'
        )),

        ('model', LinearRegression())
    ]
)
```

```python
# Hyperparameter Tuning

cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
dt_param_grid = {
    'feature_selection__k': [8, 10, 'all'],

    'model__max_depth': [None, 10, 20],

    'model__min_samples_split': [2, 10],

    'model__min_samples_leaf': [1, 5]
}
```

```python
dt_grid_search = GridSearchCV(
    estimator=decision_tree_pipeline,

    param_grid=dt_param_grid,

    cv=cv,

    scoring='neg_root_mean_squared_error',

    n_jobs=-1,

    verbose=1
)

dt_grid_search.fit(
    X_train,
    y_train
)
```

```python
print("Best Decision Tree Parameters:")

print(dt_grid_search.best_params_)

print("\nBest Cross Validation RMSE:")

print(-dt_grid_search.best_score_)
```

```python
lr_param_grid = {
    'feature_selection__k': [8, 10, 'all'],

    'model__fit_intercept': [True, False]
}
```

```python
lr_grid_search = GridSearchCV(
    estimator=linear_regression_pipeline,

    param_grid=lr_param_grid,

    cv=cv,

    scoring='neg_root_mean_squared_error',

    n_jobs=-1,

    verbose=1
)

lr_grid_search.fit(
    X_train,
    y_train
)
```

```python
print("Best Linear Regression Parameters:")

print(lr_grid_search.best_params_)

print("\nBest Cross Validation RMSE:")

print(-lr_grid_search.best_score_)
```

```python
tuning_results = pd.DataFrame({

    'Model': [
        'Decision Tree',
        'Multiple Linear Regression'
    ],

    'Best CV RMSE': [
        -dt_grid_search.best_score_,
        -lr_grid_search.best_score_
    ],

    'Best Parameters': [
        str(dt_grid_search.best_params_),
        str(lr_grid_search.best_params_)
    ]

})

display(tuning_results)
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=tuning_results,
    x='Model',
    y='Best CV RMSE'
)

plt.title('Hyperparameter Tuning - Model Comparison')
plt.xlabel('Machine Learning Models')
plt.ylabel('Cross Validation RMSE')

plt.show()
```

```python
best_decision_tree = dt_grid_search.best_estimator_

best_linear_regression = lr_grid_search.best_estimator_

print("Best Decision Tree Pipeline:")

print(best_decision_tree)

print("\nBest Linear Regression Pipeline:")

print(best_linear_regression)
```

```python
def get_selected_features(pipeline):

    preprocessor = pipeline.named_steps['preprocessor']

    selector = pipeline.named_steps['feature_selection']

    feature_names = preprocessor.get_feature_names_out()

    selected = feature_names[selector.get_support()]

    return selected


print("Decision Tree Selected Features:")

print(get_selected_features(best_decision_tree))

print("\nLinear Regression Selected Features:")

print(get_selected_features(best_linear_regression))
```

```python
def evaluate_model(y_true, y_pred):

    mae = mean_absolute_error(y_true, y_pred)

    mse = mean_squared_error(y_true, y_pred)

    rmse = np.sqrt(mse)

    r2 = r2_score(y_true, y_pred)

    return {
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2 Score': r2
    }
```
